```python
import matplotlib.pyplot as plt
import os

def plot_loss_curves(train_losses, val_losses, model_name, ticker, save_dir="reports/figures"):
    os.makedirs(save_dir, exist_ok=True)
    
    plt.figure(figsize=(8, 5))
    plt.plot(train_losses, label="Train Loss", color="blue")
    plt.plot(val_losses, label="Validation Loss", color="orange", linestyle="--")
    
    plt.title(f"Loss Curve: {model_name} ({ticker})", fontsize=12)
    plt.xlabel("Epochs")
    plt.ylabel("Loss (MSE)")
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    save_path = os.path.join(save_dir, f"{ticker}_{model_name.lower()}_loss_curve.png")
    plt.savefig(save_path, bbox_inches="tight", dpi=300)
    plt.close()
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Directory setup
REPORTS_DIR = os.path.join("..", "reports")
FIGURES_DIR = os.path.join(REPORTS_DIR, "figures")
os.makedirs(FIGURES_DIR, exist_ok=True)

# Load evaluation results from previous steps
ml_results = pd.read_csv(os.path.join(REPORTS_DIR, "ml_model_comparison.csv"))
dl_results = pd.read_csv(os.path.join(REPORTS_DIR, "dl_model_comparison.csv"))

# Combine into master summary
master_df = pd.concat([ml_results, dl_results], ignore_index=True)
master_df = master_df.sort_values(by="RMSE", ascending=True)

# Export master evaluation summary for app.py
master_df.to_csv(os.path.join(REPORTS_DIR, "master_model_comparison.csv"), index=False)
master_df
```

```python
# Function to plot and export loss curve
def plot_and_save_loss(epochs, train_loss, val_loss, model_name):
    plt.figure(figsize=(8, 4))
    plt.plot(epochs, train_loss, label="Train Loss", color="#1f77b4", linewidth=2)
    plt.plot(epochs, val_loss, label="Val Loss", color="#ff7f0e", linestyle="--", linewidth=2)
    plt.title(f"{model_name} - Training vs Validation Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss (MSE)")
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Save figure to reports/figures/ for Streamlit display
    save_path = os.path.join(FIGURES_DIR, f"{model_name.lower()}_loss_curve.png")
    plt.savefig(save_path, bbox_inches="tight", dpi=300)
    plt.close()
    print(f"Saved: {save_path}")

# Example invocation using history dicts from notebook 04
# plot_and_save_loss(range(1, 21), lstm_train_losses, lstm_val_losses, "LSTM")
```
